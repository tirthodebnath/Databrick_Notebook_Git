```python
df = spark.sql("Select * from workspace.tirtho_db.orders") 
df.display()
```

```python
from pyspark.sql import functions as F

df_with_total = df.withColumn("total_price", F.col("quantity") * F.col("price_per_unit"))
df_new = df_with_total.groupBy("order_id").agg(F.sum("total_price").alias("total_revenue")).orderBy("order_id")
display(df_new)
```

```python
from pyspark.sql import functions as F

df_with_total2 = df.withColumn("total_price", F.col("quantity") * F.col("price_per_unit"))
df_new1 = df_with_total2.groupBy("customer_id").agg(F.sum("total_price").alias("total_revenue"))
df_new3 = df_new1.orderBy(F.desc("total_revenue")).limit(3)
display(df_new3)
```
