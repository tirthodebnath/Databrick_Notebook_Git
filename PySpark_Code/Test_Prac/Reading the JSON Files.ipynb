```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Read Multiline JSON") \
    .getOrCreate()

file_path = "/Volumes/workspace/tirtho_db/tirtho_uploaded_files/*.json"

df = spark.read \
    .option("multiline", "true") \
    .option("inferSchema", "true") \
    .json(file_path)

df.printSchema()
df.display(truncate=False)
```
